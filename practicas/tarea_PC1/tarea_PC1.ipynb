Importar librerías y subir archivo .csv

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/SALINASSebas/geocomputacion_-1GEO20-/refs/heads/main/tarea_PC1/datos.csv"

df = pd.read_csv(url)
df.head()
```

### Pregunta 1:
Renombrar la columna "CÓDIGO" por "mes" y eliminar la columna "PROM. TOTAL"

```python
# Para renombrar la columna "CÓDIGO" se usará el comando .rename y se le asignará el nuevo nombre a la columna por medio de un diccionario

df = df.rename(columns = {"CÓDIGO":"mes","Día":"dia"})

# Confirmamos que se ha renombrado la columna "CÓDIGO" del dataframe
df.head()
```

```python
# Para eliminar la columna "PROM. TOTAL" se actualizará el dataframe abarcando solo los datos de las columnas deseadas mediante el comando .iloc

df = df.loc[:,"mes":"2024"]

# Confirmamos que ya no se encuentra la columna "PROM. TOTAL" en las columnas del dataframe
df.head()
```

### Pregunta 2:
Reestructurar la tabla para tener una fila por fecha real (con esto se debe eliminar los 29/02 de los años no bisiestos)

```python
# Se recolectaron los datos del dataframe dentro de una variable separada 
df_datos = df.loc[:,"1968":"2024"]

# Se crea un dataframe dentro de una variable nueva para que recopile los datos del siguiente proceso
df_nuevo = pd.DataFrame(columns=["a","b","c"])

# Con un proceso iterativo for, se recolectan los datos de mes, día y la data del año correspondiente a la columna respectova del df_datos. El nombre de las columnas del dataframe es actualizado en cada iteración para coincidir con el año de cada iteración. Se usa .concat() para apilar los datos de cada año en una sola columna consecutiva
for column in df_datos.columns:
    df_nuevo.columns = df.loc[:,["mes","dia",column]].columns
    df_nuevo = pd.concat([df_nuevo,df.loc[:,["mes","dia",column]]],sort=False)

# Las columna se renombra pues queda como el último año de data recopilada
df_nuevo = df_nuevo.rename(columns={"2024":"data"})
df_nuevo
```

```python
# El index del proceso de concatenado vuelve a 0 luego de completar la data de cada año. Se usa esto para definir una nueva lista que recopile el año de cada dato registrado
lista_año = []
año_inicio = 1968
for i in df_nuevo.index:
    lista_año.append(año_inicio)
    if i==365:
        año_inicio+=1

# Se agrega una nueva columna de año y se registran los datos de la lista creada
df_nuevo["año"] = lista_año
```

```python
# Se renombran las columnas y se visualizan los datos
df_nuevo = df_nuevo.rename(columns={"año":"year","mes":"month","dia":"day"})

df_nuevo.head()
```

```python
# Para eliminar las fechas inexistentes (29 de febrero en años no bisiestos) se transformará el formato de dato de las columnas "month", "day" y "year" a formato "datetime". Primero se crea un df nuevo con la información cronológica de cada dato.
df_fecha = df_nuevo.loc[:,["year","month","day"]]

# Se transforman los datos de las fechas "month" y "day" a tipo "int" para su posterior procesamiento
df_fecha["month"] = df_fecha["month"].astype(int)
df_fecha["day"] = df_fecha["day"].astype(int)

# El df_fecha_final guarda los datos transformados por el comando pd.to_datetime. Este recoge la data de las columnas "month", "day" y "year" y las reemplaza. El argumento "errors = coerce" permite que el comando guarde las fechas inexistentes como NaT.
df_fecha_final = pd.to_datetime(df_fecha, errors = "coerce")
df_fecha_final.head()
```

```python
# Se crea una nueva columna para la fecha y se seleccionan los datos del dataframe df_fecha_final
df_nuevo["date"] = df_fecha_final
df_nuevo.head()
```

```python
# Se eliminan las filas con fechas inexistentes con el comando .dropna en la columna "date" (toma los datos NaT generados)
df_nuevo = df_nuevo.dropna(subset=["date"])

# Se eliminan las columnas "month", "day" y "year"
df_nuevo = df_nuevo.loc[:,["date","data"]]

# Se resetea el índice del dataframe para una mejor identificación de elementos
df_nuevo = df_nuevo.reset_index(drop=True)

# Se corrige el tipo de dato de la columna "data"
df_nuevo["data"] = df_nuevo["data"].astype(float)

df_nuevo
```

### Pregunta 3
Identificar qué valores son NaN originalmente y cuáles están fuera del rango físico aceptable, guardando esa información en dos nuevas columnas: "es_dato_faltante" y "es_dato_corregido" (formato booleano, True/False)

```python
# La columna "es_dato_faltante" se crea como resultado del comando .isna() aplicado en la columna "data". El comando devuelve un set de valores "True" y "False" si el valor del elemento de la columna "data" es NaN o no, respectivamente (PASO REALIZADO CON AYUDA DE IA)
df_nuevo["es_dato_faltante"] = df_nuevo["data"].isna()

# La columna "es_dato_corregido" se crea tomando valores "False" bajo el supuesto de que ningún dato ha sido modificado
df_nuevo["es_dato_corregido"] = False

df_nuevo.head()
```

### Pregunta 4
Reemplazar los valores fuera de rango por NaN.

```python
# Los comandos "max" y "min" permiten ver los valores máximos y mínimos del set de datos de la columna "data". Se usa el comando .dropna() para no incluir los valores NaN dentro del comando "max" y "min"
print(max(df_nuevo["data"].dropna()))
print(min(df_nuevo["data"].dropna()))
```

```python
# Debido a que una temperatura de 218.0 para representar el clima no presenta sentido físico, se identifica como el dato que debe ser reemplazado. Para eso, se determina la ubicación del dato en el dataframe.
df_nuevo.loc[df_nuevo["data"]==218]
```

```python
# El dato es reemplazado con el comando .replace() a un valor nulo (NaN)
df_nuevo = df_nuevo.replace(218,np.nan)

# Luego de haber procesado el dato fuera de rango, se vuelve a identificar el valor máximo de la columna "data"
print(max(df_nuevo["data"].dropna()))
```

```python
# Debido a que el valor fuera de rango ha sido corregido, se registra la modificación en la columna "es_dato_corregido"
df_nuevo.loc[17606,"es_dato_corregido"] = True

df_nuevo.loc[17606,:]
```

### Pregunta 5
Agregar la columna "variable" (la variable debe estar en formato texto: str, sin espacios ni tildes)

```python
# Se crea una nueva columna que refleje la variable analizada en el dataframe (temperatura mínima). Esta debe estar en formato "str"
df_nuevo["variable"] = "tmin"

df_nuevo.info()
```

Exportar el dataframe modificado como archivo .csv

```python
# Finalmente, se guarda el dataframe modificado en un archivo .csv con el comando .to_csv. El atributo "index = False" se usa para no guardar el índice como columna en el archivo .csv.
df_nuevo.to_csv('tmin_corregido_Sebastian.csv', index=False)
```
