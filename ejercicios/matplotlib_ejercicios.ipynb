# Ejercicios con matplotlib y pandas


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = 'https://raw.githubusercontent.com/SALINASSebas/geocomputacion_-1GEO20-/refs/heads/main/tarea_PC1/tmin_corregido_Sebastian.csv'

df = pd.read_csv(url)
df.info()
```

```python
df['date'] = pd.to_datetime(df['date'])
df.info()
```

1. Abrir el archivo de datos corregidos para una variable y graficarla. Luego cambiar figsize a 2-3 valores distintos.

```python
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(df['data'])
ax.set_title('Tmin')
```

```python
fig, ax = plt.subplots(figsize=(6,4))
ax.plot(df['data'])
ax.set_title('Tmin')
```

```python
fig, ax = plt.subplots(figsize=(8,1))
ax.plot(df['data'])
ax.set_title('Tmin')
```

2. Crear una figura con 2 paneles. En cada panel graficar una año distinto, cada uno con su propio título.

```python
fig, axes = plt.subplots(1,2, figsize=(10,4))

data_1990 = df[df['date'].dt.year == 1990]
data_2010 = df[df['date'].dt.year == 2010]

axes[0].plot(data_1990['data'])
axes[0].set_title('Tmin 1990')

axes[1].plot(data_2010["data"])
axes[1].set_title('Tmin 2010')
```

3. Graficar la distribución de una misma variable usando boxplot o violinplot.

```python
df['month'] = df['date'].dt.month

df.boxplot(column='data', by='month')
```

```python
import seaborn as sns

sns.violinplot(data=df, x='month', y='data')
```
